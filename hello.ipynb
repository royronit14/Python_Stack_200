```python
class DSU:
    def __init__(self, n):
        # Elements are numbered 0 through n - 1.
        self.parent = list(range(n))
        self.size = [1] * n

    def find_parent(self, node):
        if self.parent[node] != node:
            self.parent[node] = self.find_parent(self.parent[node])
        return self.parent[node]

    def union_by_size(self, u, v):
        root_u = self.find_parent(u)
        root_v = self.find_parent(v)

        if root_u == root_v:
            return root_u

        # Attach the smaller component's root to the larger one's root.
        if self.size[root_u] < self.size[root_v]:
            root_u, root_v = root_v, root_u

        self.parent[root_v] = root_u
        self.size[root_u] += self.size[root_v]
        return root_u
```

# Variables

Everything in python is an object.

Say a = 5 and b = 5. If we find the address of a and b, the addresses will be same for both...

```python
a = 5
b = 5

print(a)
print(b)
print()
print(f"Address of a is: {id(a)}")
print(f"Address of b is: {id(b)}")
```

So yahan pe saare variables acts like references, toh yesbb 5 (which is an object) usko refer kr rha hai

```python
b = 6
print(f"Address of b is: {id(b)}")

k = 9
print(f"Address of k is: {id(k)}")
```

```python
b = 9
print(f"Address of b is: {id(b)}")
```

Toh question: whats happens to the value b = 6.

Ans: In python there is garbage collection, it will see that object (i.e. 6, which was earlier refered from b is now not being refered) will be removed.

```python
# So object is everything, ur list, string, numbers(int), etc


# Below when using this is refered as: string interning
name1 = "ronit"
name2 = "ronit"

print(id(name1))
print(id(name2))
print()
# String interning is only happening with few words, not like long sentences, heres how:
line1 = "Hello, this is my python series"
line2 = "Hello, this is my python series"

print(id(line1))
print(id(line2))

# We see that line1 and line2 are same but unki address same ni hai
# This same thing happens when we use large numbers

n = 1000
m = 1000

print()
print(id(n))
print(id(m))
```
