# Encapsulation



Unlike languages such as Java or C++, Python does not have strict public or private keywords. Instead, it relies on naming conventions:

Public: self.name (Accessible from anywhere).

Protected (Convention): self._name (A single leading underscore signals to other developers: "This is for internal use, please do not touch this directly").

Private (Name Mangling): self.__name (Double leading underscores tell Python to actively rename the variable behind the scenes to make it difficult to access accidentally).

```python
class ModelConfig:
    def __init__(self, learning_rate):
        # We use a single underscore to indicate this is internal
        self._lr = learning_rate
    def set_learning_rate(self, new_lr):
        # The method acts as a gatekeeper to protect the data
        if new_lr <= 0:
            print("Error: Learning rate must be greater than 0.")
        else:
            self._lr = new_lr

    def get_learning_rate(self):
        return self._lr

config = ModelConfig(0.001)
# config.get_learning_rate()

# config.set_learning_rate(-0.5)
# config.get_learning_rate()


config.set_learning_rate(0.5)
config.get_learning_rate()
```

```python
class TrainingEpochs:
    def __init__(self, epoch_count = 10):
        self._epochs = epoch_count
    def set_epochs(self, amount):
        if amount < 0:
            print("Negative value isn't accepted.")
        else:
            self._epochs += amount
    def get_epochs(self):
        return self._epochs

obj = TrainingEpochs()
obj.get_epochs()

# obj.set_epochs(-5)
# obj.get_epochs()

obj.set_epochs(5)
obj.get_epochs()

```

# Polymorphism and Method Overriding

Polymorphism means "many forms." In Python, it allows different child classes to use methods that share the exact same name but behave differently.

This is arguably the most important OOP concept for PyTorch. Every single neural network layer you will ever build must inherit from a parent class called nn.Module. That parent class expects you to override a specific method called forward(). Polymorphism ensures that PyTorch can blindly call .forward() on any layer, trusting that your specific layer knows how to compute its own math.

```python
class Parent:
    def process_data(self):
        print("Basic processing")

class Child(Parent):
    # This completely overrides the parent's version
    def process_data(self):
        print("Advanced child processing")
```

```python
class Layer:
    def forward(self, input_value):
        return input_value
class RELULayer(Layer):
    def forward(self, input_value):
        if input_value < 0:
            return 0
        else:
            return input_value
class DoubleLayer(Layer):
    def forward(self, input_value):
        return input_value*2

relu = RELULayer()
doubler = DoubleLayer()

print(f"ReLU Output: {relu.forward(-5)}")
print(f"DoubleLayer Output: {doubler.forward(-5)}")
```

If you are ever confused about which version of a method is actually running, you can inspect the Method Resolution Order (MRO). Every class has a special __mro__ attribute (or a .mro() method) that returns a tuple showing the exact order Python searches for methods.
print(ReLULayer.__mro__) outputs:
(<class '__main__.ReLULayer'>, <class '__main__.Layer'>, <class 'object'>)
This proves Python checks ReLULayer before Layer.

# super() and Class Hierarchy Initialization

When a child class overrides a method (especially the __init__ constructor), it completely replaces the parent's version. But what if you don't want to replace the parent's behavior, but rather add to it?

The super() function allows a child class to temporarily act like its parent, letting you call parent methods directly from within the child.

Imagine you are building a House. The parent class Building knows how to lay a foundation and build walls. The child class House knows how to add bedrooms and a kitchen.

If House writes its own __init__, it forgets how to build a foundation. It just tries to float bedrooms in the air.

By calling super().__init__() inside the House's constructor, the House is telling the Building: "Hey parent, do your foundation and walls job first." Once the parent finishes, the child takes over and adds the bedrooms. You get the best of both worlds.

```python
class Parent:
    def __init__(self, name):
        self.name = name

class Child(Parent):
    def __init__(self, name, age):
        # 1. Call parent's __init__ to handle 'name'
        super().__init__(name) 
        
        # 2. Handle child-specific initialization
        self.age = age
```

```python
class BaseLayer:
    def __init__(self, layer_id):
        print("This is the parent class __init__ running")

        self.id = layer_id
        self.is_active = True
        print("The Parent Class __init__ is finished")

class LinearLayer(BaseLayer):
    def __init__(self, layer_id, weight_size):
        print("This is the Child Class __init__ running")

        super().__init__(layer_id)         #Parent requires 'layer_id'
        self.weights = weight_size
        print("Linear Layer __init__ finished")

layer3 = LinearLayer(101, 512)

print(f"Layer ID: {layer3.id}")
print(f"Weights of the layer: {layer3.weights}")
print(f"Is Active: {layer3.is_active}")
```
